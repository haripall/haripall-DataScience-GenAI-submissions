# Seminar 2 Task
Completing EDA on the titanic.csv data as provided in the seminar during week 3.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Uploading the dataset directly via code
print("1. Uploading the Dataset")
upload = files.upload()

# Loading the uploaded dataset as provided in the seminar - titanic.csv
print("\n2. Loading the Dataset")
df = pd.read_csv('titanic.csv')
print("Dataset 'titanic.csv' loaded successfully.")

# Checking for missing values
print("\n3. Checking for Missing Values")
missing = df.isnull().sum()
print("Missing Values:\n", missing[missing > 0])

# Checking for outliers and extreme values (Descriptive Stats)
print("\n4. Descriptive Stats (Outliers and Extreme Values)")
print(df[['Age', 'Fare']].describe())

# Univariate and Bivariate Analysis Visualizations
print("\n5. Univariate and Bivariate Analysis Visualizations")
plt.figure(figsize=(12, 10))

# Univariate: Survival
plt.subplot(2, 2, 1)
sns.countplot(data=df, x='Survived')
plt.title('Univariate: Survival Count')

# Univariate: Age
plt.subplot(2, 2, 2)
sns.histplot(data=df, x='Age', bins=30, kde=True)
plt.title('Univariate: Age Distribution')

# Bivariate: Survival by Sex
plt.subplot(2, 2, 3)
sns.barplot(data=df, x='Sex', y='Survived')
plt.title('Bivariate: Survival Rate by Sex')

# Bivariate: Survival by Ticket Class
plt.subplot(2, 2, 4)
sns.barplot(data=df, x='Pclass', y='Survived')
plt.title('Bivariate: Survival Rate by Class')

plt.tight_layout()
plt.show()

# View the first 5 rows of the dataset
print("\n6. First 5 Rows of the Dataset")
print(df.head())

# Check the overall dimensions (rows, columns) of the dataframe
print("\n7. Overall Dimensions (Rows, Columns)")
print(df.shape)

# Display column data types and non-null counts
print("\n8. Column Data Types and Non-Null Counts")
df.info()

# Calculate the percentage of missing data per column
print("\n9. Percentage of Missing Data Per Column")
print((df.isnull().sum() / len(df)) * 100)

# Check for and count any duplicate rows in the dataset
print("\n10. Duplicate Rows Count")
print(df.duplicated().value_counts())

# Create a base matplotlib bar chart of Pclass vs Fare
print("\n11. Bar Chart: Ticket Class vs Fare Paid")
plt.bar(df['Pclass'], df['Fare'])
plt.xlabel('Ticket Class')
plt.xticks([1, 2, 3]) # Ensure only valid classes show on x-axis
plt.ylabel('Fare Paid')
plt.title('Ticket Class vs Fare Paid')
plt.show()

# Create a base matplotlib scatter plot of Age vs Fare
print("\n12. Scatter Plot: Age vs Fare Paid")
plt.scatter(df['Age'], df['Fare'])
plt.xlabel('Age')
plt.ylabel('Fare Paid')
plt.title('Age vs Fare Paid')
plt.show()

# Create a distribution plot for Fare specifically
print("\n13. Distribution Plot for Fare")
sns.displot(df['Fare'])
plt.show()

# Generate a pairplot to compare multiple numeric variables simultaneously
print("\n14. Pairplot: Age, Fare, and Pclass")
sns.pairplot(data=df[['Age', 'Fare', 'Pclass']], diag_kind='hist')
plt.show()

# Drop columns that are unneeded or have too much missing data
print("\n15. Dropping 'Cabin' and 'Ticket' Columns")
df = df.drop(['Cabin', 'Ticket'], axis=1)
print(df.head())
```

# Surprising insights from the data
The "Women and Children First" rule was strongly enforced by class. Females in 1st class had a near-perfect survival rate of 96.8%, and females in 2nd class survived at a 92.1% rate. However, females in 3rd class only had a 50% survival rate—meaning a 3rd class woman was almost as likely to perish as to survive.

Family size heavily dictated survival odds. Passengers traveling alone had a poor survival rate of 30.3%. Traveling with a small family (2-4 people total) drastically improved survival chances, peaking at 72.4% for families of 4. However, having a large family (5 or more people) was fatal; survival rates plummeted back down, with 0% survival for families of 8 and 11.

Embarkation port mattered. Passengers who boarded at Cherbourg (C) had a significantly higher survival rate (55.3%) compared to those who boarded at Southampton (S) (33.7%) or Queenstown (Q) (38.9%).

A male in 1st class was still at high risk. Despite the advantages of wealth, 1st class males only had a 36.8% survival rate, which was substantially lower than the survival rate of 3rd class females (50%).
