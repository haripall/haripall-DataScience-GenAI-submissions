First, I made a DataFrame from a dictionary. Each dictionary key becomes a column name, and each list supplies the values in that column.

```python
import pandas as pd

products = {
    'product': ['Hoodie', 'T-shirt', 'Tote bag', 'Pencil'],
    'price_gbp': [28.00, 18.00, 8.50, 1.50],
    'in_stock': [True, True, False, True]
    }

products_df = pd.DataFrame(products)
products_df
```

A DataFrame can also be made from a list of dictionaries. Each dictionary represents one row. If every row has the same fields, the result is a rectangular table.


```python
import pandas as pd

sales = [
    {'order_id': 'A101', 'product': 'Hoodie', 'quantity': 1},
    {'order_id': 'A102', 'product': 'Tote bag', 'quantity': 2},
    {'order_id': 'A103', 'product': 'Pencil', 'quantity': 10}
]

sales_df = pd.DataFrame(sales)
sales_df
```

When one item is missing: a and b have four items, but c_short has only three. I caught the expected error so that the notebook can continue running with a clear result

```python
import pandas as pd
a = [1, 2, 3, 4]
b = ['a', 'b', 'c', 'd']
c_short = [True, False, True]

try:
    unequal_rows_df = pd.DataFrame([a, b, c_short])
except ValueError as error:
    print('ValueError raised because the rows have different lengths.')
    print(error)
```

The DataFrame is not created from those three unequal length lists. Pandas needs a rectangular table, so the rows in this list-of-lists input must have the same number of values. It does not automatically guess the missing fourth value. If a value is missing, keep each column the same length and use None for the unknown entry. Pandas represents that missing value as NaN in this numeric column.

```python
import pandas as pd

data_with_missing_value = {
    'number': [1, 2, 3, 4],
    'letter': ['a', 'b', 'c', 'd'],
    'flag': [True, False, True, None]
}

missing_value_df = pd.DataFrame(data_with_missing_value)
missing_value_df
```

Another pattern is a list of dictionaries. If one row omits a field, Pandas fills that cell with a missing-value marker (NaN/None) rather than rejecting the whole DataFrame.

```python
import pandas as pd

records_with_missing_field = [
    {'number': 1, 'letter': 'a', 'flag': True},
    {'number': 2, 'letter': 'b', 'flag': False},
    {'number': 3, 'letter': 'c', 'flag': True},
    {'number': 4, 'letter': 'd'}  # flag was not recorded for this row
]

records_df = pd.DataFrame(records_with_missing_field)
records_df
```

# Conlusion
A Pandas DataFrame stores structured data in rows and columns, similar to a spreadsheet.

A dictionary of equally sized lists is a convenient way to define columns.

A list of dictionaries is a convenient way to define rows.

When creating a DataFrame from a list of lists, the lists need equal lengths or else a ValueError will occur.

Missing data can instead be represented explicitly (for example with None), or by omitting a key in one dictionary row. Resulting missing cells will show as NaN where applicable

### What are Dataframes?
Pandas has a very elegant way of managing data, very much borrowed from the statistical language R, mostly based around dataframes. You can think of dataframes a bit like an Excel table with rows, columns and common operations like sum, average and so on. We can use this on top of our previous work on lists and dictionaries (etc.) more usable and malleable.

To begin with we will create a dataset - in this case a [dictionary](https://github.com/MJMortensonWarwick/IB2AD0_Data_Science_GenerativeAI/blob/main/1_05_dictionaries.ipynb).



```python
orders = {'o10001':{'date':'2024/01/10', 'product':'Hoodie', 'quantity':'1'},
            'o10002':{'date':'2024/01/13', 'product':'Tote bag', 'quantity':'2'},
            'o10003':{'date':'2024/01/14', 'product':'Pencil', 'quantity':'10'},
            'o10004':{'date':'2024/01/15', 'product':'T-shirt', 'quantity':'2'}
}
orders
```

We can convert this to a Dataframe with great ease

```python
import pandas as pd
import numpy as np

orders_df = pd.DataFrame(orders)
orders_df
```

We can even create such outputs from more complex dictionaries, such as a dictionary which includes a nested list:

```python
customers = {'Mark':{'name':'Mark Johnson', 'open_orders':3, 'orders':['o10001', 'o10002', 'o10004']},
             'Katy':{'name':'Katy Hoad', 'open_orders':0, 'orders':[]},
             'Angela':{'name':'Angela Lorenz', 'open_orders':1, 'orders':['o10003']},
             'Bo':{'name':'Bo Kelestyn', 'open_orders':0, 'orders':[]}
}
customers
```

```python
customers_df = pd.DataFrame(customers)
customers_df
```

Pandas dataframes can also be built from various other data structures and sources (including Excel files, CSVs, text files, databases and many more). For example, we can make our dataframe from a set of [lists](https://github.com/MJMortensonWarwick/IB2AD0_Data_Science_GenerativeAI/blob/main/1_04_lists.ipynb):

```python
a = [1, 2, 3, 4]
b = ["a", "b", "c", "d"]
c = [True, False, True, False]

listdf = pd.DataFrame([a, b, c])
listdf
```
